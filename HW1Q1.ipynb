```python
import random
from itertools import permutations

def secret_santa(participants, imperm=[]):

    original = participants[:]
    valid_pairs = []

    # Loop through all permutations
    for p in permutations(original):
        valid = True
        # Check for impermissible pairs
        for i, giver in enumerate(p):
            receiver = p[(i + 1) % len(p)]  # Last person gives to first
            if (giver, receiver) in imperm:
                valid = False
                break

        # Add this pair to the results
        if valid:
            pairs = []
            for i, giver in enumerate(p):
                receiver = p[(i + 1) % len(p)]
                pairs.append((giver, receiver))
            valid_pairs.append(pairs)

    return valid_pairs


participants = ["Jim", "Tim", "Kim", "Jane", "John"]
impermissible = [("Jim", "John")] #Impermissable pair

# Add list to the function
all_combs = secret_santa(participants, impermissible)

# Print combinations
if all_combs:
    print("All combinations:")
    for idx, pairs in enumerate(all_combs, start=1):
        for giver, receiver in pairs:
            print(f"{giver} : {receiver}")
        print()


#Citations

# https://binary-machinery.github.io/2021/02/03/secret-santa-graph.html :
#Approach for secret santa based on this article, where participants are arranged in a list.
#Each participant is assigned a random receiver from the remaining participants, and the chain is closed by having the last participant give a gift to the first.

#ChatGPT (OpenAI, 2025), response to prompt: "How to handle the edge case where the last person in the list must give to the first person?”"
#https://www.w3schools.com/python/python_ref_functions.asp for syntax and examples
```
